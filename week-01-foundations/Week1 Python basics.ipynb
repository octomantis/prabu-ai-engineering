```python
class LearningTopic:
    def __init__(self, name, week):
        self.name = name
        self.week = week
        self.completed = False
```

```python
python_topic = LearningTopic("Python", 1)
ml_topic = LearningTopic("Machine Learning", 1)
```

```python
print (f"{python_topic.name} - name, {python_topic.week} - Week, {python_topic.completed}")
print (f"{ml_topic.name} - name, {ml_topic.week} - week, {ml_topic.completed}")
```

```python
python_topic.completed = True
```

```python
print (f"{python_topic.name} - name, {python_topic.week} - Week, {python_topic.completed}")
print (f"{ml_topic.name} - name, {ml_topic.week} - week, {ml_topic.completed}")
```
